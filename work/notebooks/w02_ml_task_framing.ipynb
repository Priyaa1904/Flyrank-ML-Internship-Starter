### Unit of analysis

The unit of analysis is one content page. Each row represents a single page identified by `content_id`, with page-level search, engagement, freshness and performance signals. The raw starter data does not contain a direct `should_refresh` target, so I sketch `is_declining_proxy` from `trend_direction` as a possible proxy target. A value of 1 represents a page with a `down` trend and 0 represents other trend categories. This proxy would support prioritization rather than automatically deciding that a page must be refreshed.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule is useful as a simple baseline but it would require manually choosing thresholds and combinations of signals. For example, a rule based only on low CTR could prioritize pages that have low CTR but little search visibility or other reasons for their performance. An ML ranking approach can learn patterns across multiple page-level signals and assign a relative priority score to pages. This could make the prioritization more flexible than a single fixed threshold. However, ML should only be preferred if it demonstrates better performance on an honest validation setup than the fixed-rule baseline.
